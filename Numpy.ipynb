```python
#Views VS Copies
arr = np.array([1, 2, 3, 4, 5])

# Creating a view
view_arr = arr.view()
print("Original array:", arr)
print("View array:", view_arr)

# Modifying the view affects the original array
view_arr[0] = 100
print("\nAfter modifying view_arr[0] = 100:")
print("Original array:", arr)
print("View array:", view_arr)

# Creating a copy
copy_arr = arr.copy()
print("\nOriginal array (after view modification):", arr)
print("Copy array:", copy_arr)

# Modifying the copy does NOT affect the original array
copy_arr[1] = 200
print("\nAfter modifying copy_arr[1] = 200:")
print("Original array:", arr)
print("Copy array:", copy_arr)
```

```python
#broadcasting

# Broadcasting with a scalar
arr_scalar = np.array([1, 2, 3])
scalar = 2
result_scalar = arr_scalar * scalar
print("Array multiplied by scalar:", result_scalar)

# Broadcasting with arrays of different shapes
arr_a = np.array([[1, 2, 3], [4, 5, 6]]) # Shape (2, 3)
arr_b = np.array([10, 20, 30])         # Shape (3,)

result_array = arr_a + arr_b
print("\nArrays added with broadcasting:")
print("arr_a (shape 2,3):\n", arr_a)
print("arr_b (shape 3,):", arr_b)
print("Result:\n", result_array)

# Another example: column vector broadcasting
arr_c = np.array([[100], [200]]) # Shape (2, 1)
result_col_vec = arr_a + arr_c
print("\nArray added with column vector broadcasting:")
print("arr_a (shape 2,3):\n", arr_a)
print("arr_c (shape 2,1):\n", arr_c)
print("Result:\n", result_col_vec)
```

```python
#Vectorization

import time

# Non-vectorized approach (using a loop)
list_a = list(range(1000000))
list_b = list(range(1000000))
result_list = []

start_time = time.time()
for i in range(len(list_a)):
    result_list.append(list_a[i] + list_b[i])
end_time = time.time()
print(f"Time taken for non-vectorized operation: {end_time - start_time:.6f} seconds")

# Vectorized approach (using NumPy arrays)
numpy_a = np.array(range(1000000))
numpy_b = np.array(range(1000000))

start_time = time.time()
result_numpy = numpy_a + numpy_b
end_time = time.time()
print(f"Time taken for vectorized operation: {end_time - start_time:.6f} seconds")

print("\nFirst 5 elements of non-vectorized result:", result_list[:5])
print("First 5 elements of vectorized result:", result_numpy[:5])
```

```python
#isnan,isinf,isfinite

import numpy as np

arr_with_special_values = np.array([1, 0, np.nan, np.inf, -np.inf, 5])
print("Original array:", arr_with_special_values)

# Check for NaN (Not a Number)
print("Is NaN:", np.isnan(arr_with_special_values))

# Check for positive or negative infinity
print("Is Infinite:", np.isinf(arr_with_special_values))

# Check for finite values (not NaN and not infinity)
print("Is Finite:", np.isfinite(arr_with_special_values))
```
