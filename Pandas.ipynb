```python
#Dataframes
data={
    "Name":["Alice", "Bob", "Charlie", "David", "Eve", "Frank", "Grace"],
    "Age":[25, 30, np.nan, 35, 40, 28, np.nan],
    "Department":["HR", "IT", "Finance", np.nan, "Marketing", "HR", "IT"],
    "Salary":[50000.0, 60000.0, np.nan, 80000.0, 70000.0, 55000.0, 62000.0]
}
df = pd.DataFrame(data)
df
```

```python
df.iloc[1:3]
```

```python
df.loc[1:3,["Age","Department"]]
```

```python
#Drop Columns
df.drop("Age",axis=1)
```

```python
df
```

```python
df.drop("Age",axis=1,inplace=True)
```

```python
df
```

```python
df.shape
```

```python
df.info()
```

```python
df.describe()
```

```python
#Broadcasting
df['Salary']=df["Salary"]+5000
df['Salary']
```

```python
#renaming columns
df.rename(columns={"Department":"Dept"},inplace=True)
```

```python
df
```

```python
#Unique values
df["Dept"].unique()
```

```python
#value_counts
df["Dept"].value_counts()
```

```python
#Add new column
df["Promoted Salary"]=df["Salary"]*10
```

```python
df
```

```python
#Data Cleaning
df.isnull()
```

```python
df.isnull().sum()
```

```python
#drop null values
df.dropna()
```

```python
df.dropna(how="any")
```

```python
df.dropna(how="all")
```

```python
#fill the null values
df.fillna(0)
```

```python
df["Salary"].fillna(df["Salary"].mean())
```

```python
#forward fill
df["Dept"].fillna(method="ffill")
```

```python
#backward fill
df["Dept"].fillna(method="bfill")
```

```python
#replace
df["Name"].replace("Alice","Rose")
```

```python
#Duplicates
df_dup=df[df.duplicated()]
df_dup
```

```python
#drop duplicates
df=df.drop_duplicates()
```

```python
#invalid values
df['Promoted Salary']=df['Promoted Salary'].apply(lambda x: x/10 if x>6000 else x)
```

```python
df['Promoted Salary']
```

```python
df
```

```python
#Joins and merge
department_info={
    "Department":["HR","IT","Finance","Marketing"],
    "Location":["New York","San Francisco","Los Angeles","Chicago"],
    "Manager":["Alice","Bob","Charlie","David"]
}
```

```python
df2=pd.DataFrame(department_info)
df2
```

```python
#Merge two datasets
pd.concat([df,df2])
```

```python
#column wise
pd.concat([df,df2],axis=1)
```

```python
pd.merge(df, df2, left_on="Dept", right_on="Department")
```
